### Problema 1:
Sacar el total, promedio y ventas altas de una lista.

```python
ventas = [100, 300, 150, 400, 250] #Lista de ventas

total = 0 #Variable para guardar el total de ventas

# Ciclo que suma cada venta y obtiene el total
for venta in ventas:
    total = total + venta

#Operación para obtener el promedio de ventas
promedio = total / len(ventas)

#Datos obtenidos
print("El total es:", total)
print("El promedio es:", promedio)

#Ciclo que indica las ventas altas (>=250) de la lista
for venta in ventas:
    if venta >= 250:
        print("Ventas altas: ", venta)
    else:
        print("Ventas bajas: ", venta)
```
